```python
import matplotlib.pyplot as plt
```

```python
x=[1,2,3,4,5]
y=[10,22,35,45,50]
plt.plot(x,y)
plt.show()
```

```python
plt.plot(x,y,color='red',marker='o',linestyle='--',linewidth=1,markersize=2)
plt.show()
```

```python
plt.figure(figsize=(6,2))
plt.plot(x,y,color='red',marker='o',linestyle='--',linewidth=1,markersize=2)
plt.title("line plot")
plt.xlabel("x-axis")
plt.ylabel("y-axis")

plt.show()
```

```python
plt.figure(figsize=(6,6))
plt.plot(x,y,color='red',marker='o',linestyle='--',linewidth=1,markersize=2)
plt.title("line plot")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.grid(True)
plt.show()
```

```python
x = [1,2,3,4,5]
y = [10, 20, 25, 35, 45]
y1=[34,35,38,56,78]
plt.plot(x,y,label='sales of 2024')
plt.plot(x,y1,label='sales of 2025')
plt.title("line plot")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.legend()
plt.show()
```

```python
x = ['A','B','C','D','E']
y = [10, 20, 25, 35, 45]
plt.bar(x,y)
plt.title("bar plot")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.show()
```

```python
import random
data=[random.randint (1,20) for _ in range(100)]

plt.hist(data,bins=5)
plt.title("hist plot")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.show()
```

```python
category = ['A','B','C','D','E']
sales = [10, 20, 25, 35, 45]
plt.pie(sales,labels=category,autopct='%1.1f%%',startangle=90)
plt.title("pie plot")
plt.show()
```

```python
y = [10, 20, 25, 35, 45]
y1=[34,35,38,56,78]
plt.scatter(y,y1)
plt.title("scatter plot")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.show()
```

```python
category = ['A','B','C','D','E']
sales = [10, 20, 25, 35, 45]

y = [10, 20, 25, 35, 45]
y1=[34,35,38,56,78]

plt.subplot(2,1,1)
plt.bar(category,sales)
plt.title("bar plot")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.figure(figsize=(2,2))
plt.subplot(2,1,2)
plt.scatter(y,y1)
plt.title("scatter plot")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.show()
```

```python
import pandas as pd
data={
    'month':['jan','feb','mar','apr','may'],
    'sales':[100,200,300,400,500]
}
df=pd.DataFrame(data)
print(df)
```

```python
plt.bar(df['month'],df['sales'])
plt.title("pandas")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.show()
```
